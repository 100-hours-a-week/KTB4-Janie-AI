### **고급 다중 그래프 - 미니퀘스트**

`plt.subplots()`를 사용하여 **2x1 형태의 서브플롯을 만들고**, 첫 번째 서브플롯에는 `y = x^2`, 두 번째 서브플롯에는 `y = x^3`을 그리는 코드를 작성하세요.

```python
x = np.linspace(0, 10, 100)
y1 = x*x
y2 = x*x*x

fig, axes = plt.subplots(nrows=2, ncols=1, figsize=(8, 6))

axes[0].plot(x, y1, color='b', label='x^2')
axes[0].set_xlabel('X-axis')
axes[0].set_ylabel('Y-axis')
axes[0].legend()

axes[1].plot(x, y2, color='y', label='x^3')
axes[1].set_xlabel('X-axis')
axes[1].set_ylabel('Y-axis')
axes[1].legend()
```

```python
import matplotlib.gridspec as gridspec

fig = plt.figure(figsize=(10, 8))
gs = gridspec.GridSpec(3, 2, figure=fig)

ax1 = fig.add_subplot(gs[0, :])
ax1.plot(x, y1, color='b')
ax1.legend()

ax2 = fig.add_subplot(gs[1, 0])
ax2.scatter(x, y1, color='g')
ax2.legend()

ax3 = fig.add_subplot(gs[1, 1])
ax3.bar(categories, values, color='g')
ax3.legend()

ax4 = fig.add_subplot(gs[2,:])
ax4.hist(y2)

plt.tight_layout()
plt.show()
```

```python
intersection = len(set_A & set_B)
diagram = venn2([set_A, set_B], set_labels=('Set A', 'Set B'))

if intersection >= 2:
  color='blue'
else:
  color='lightgreen'

intersection_area = diagram.get_patch_by_id('11')

if intersection_area:
  intersection_area.set_color(color)

plt.show()
```

이상치는 사분위수 범위(IQR)를 이용해 판단합니다.

</aside>

```python
# 시계열 데이터 생성
np.random.seed(42)
date_range = pd.date_range(start="2023-01-01", periods=100, freq="D")
values = np.cumsum(np.random.randn(100))
```

```python
Q1 = df['Value'].quantile(0.25)
Q3 = df['Value'].quantile(0.75)
IQR = Q3 - Q1
k = 0.5
lower_bound = Q1 - k * IQR
upper_bound = Q3 + k * IQR

# 이상치 판별
df['Outlier'] = (df['Value'] < lower_bound) | (df['Value'] > upper_bound)
outliers = df[df['Outlier']]

plt.figure(figsize=(10, 5))
sns.lineplot(x='Date', y='Value', data=df, label='Original Data', color='navy')
sns.scatterplot(x='Date', y='Value', data=outliers, color='orange', label='Outliers', s=100)

plt.xlabel('Date')
plt.ylabel('Value')
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

**데이터에서 이상값(Outlier)을 찾아 제거한 후, 원래 데이터와 이상값 제거 후 데이터의 평균을 비교하는 코드**를 작성하세요.

이상값은 **IQR(사분위 범위)를 사용하여 탐지**하세요.

</aside>

```python
# 샘플 데이터 생성
np.random.seed(42)
data = np.random.normal(loc=50, scale=10, size=100)  # 평균 50, 표준편차 10인 정규 분포 데이터 생성
df = pd.DataFrame(data, columns=["value"])  # 데이터프레임 생성
```

```python
q1 = np.percentile(df["value"], 25)
q3 = np.percentile(df["value"], 75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
outliers = (df["value"] < lower_bound) | (df["value"] > upper_bound)
df_copy = df.copy()
df_copy = df[~outliers]
print(df['value'].mean() - df_copy['value'].mean())
```

**데이터의 왜도(Skewness)와 첨도(Kurtosis)를 계산하여 데이터의 분포 특성을 분석하는 코드**를 작성하세요.

</aside>

```python
# 샘플 데이터 생성
np.random.seed(42)
data = np.random.normal(loc=50, scale=10, size=100)  # 평균 50, 표준편차 10인 정규 분포 데이터 생성
df = pd.DataFrame(data, columns=["value"])  # 데이터프레임 생성
```

```python
skewness = stats.skew(df["value"])
kurtosis = stats.kurtosis(df["value"])

print("왜도:", skewness)
print("첨도:", kurtosis)
```

```python
# 왼쪽으로 치우침, 평평한 분포
if skewness > 0:
    print("오른쪽으로 꼬리가 긴 분포")
elif skewness < 0:
    print("왼쪽으로 꼬리가 긴 분포")
else:
    print("대칭에 가까운 분포입")

if kurtosis > 0:
    print("정규분포보다 뾰족하고 꼬리가 두꺼운 분포")
elif kurtosis < 0:
    print("정규분포보다 완만하고 꼬리가 얇은 분포")
else:
    print("정규분포와 비슷한 첨도")
```

### **가설 검정 - 미니퀘스트**
1번 퀘스트

<aside>
👉

**단일 표본 t-검정(One-Sample t-test)** 을 수행하여 샘플 데이터의 평균이 특정 값과 유의미한 차이가 있는지 검정하는 코드를 작성하세요.

(평균 `50`, 표준 편차 `5`를 따르는 정규 분포에서 30개의 데이터를 생성하고, 해당 데이터가 평균 `52`와 차이가 있는지 확인하세요.)

</aside>

```python
from scipy import stats
np.random.seed(42)
group_A = np.random.normal(loc=50, scale=5, size=30)
t_stat, p_value = stats.ttest_1samp(group_A, 52)
print(f"t-통계량: {t_stat}")
print(f"p-value:{p_value}")
```

2번 퀘스트

<aside>
👉

**카이제곱 검정(Chi-Square Test)** 을 수행하여 관측된 데이터와 기대값이 유의미한 차이가 있는지 확인하는 코드를 작성하세요.

</aside>

```python
# 관측된 데이터 (Observed)
observed = np.array([50, 60, 90])

# 기대값 (Expected)
expected = np.array([66, 66, 66]) * (observed.sum() / np.sum([66, 66, 66]))
```

```python
t_stat, p_value = stats.chisquare(observed, expected)
if p_value < 0.05:
  print(f"p_value: {p_value}, 통계적으로 유의미한 차이")
else:
  print("p_value: {p_value}, 유의미한 차이 X")
```

3번 퀘스트

<aside>
👉

**분산 분석(ANOVA, Analysis of Variance)** 을 수행하여 여러 그룹의 평균이 서로 다른지 검정하는 코드를 작성하세요.

</aside>

```python
# 샘플 데이터 생성
np.random.seed(42)
group_1 = np.random.normal(loc=50, scale=10, size=30)  # 평균 50, 표준편차 10
group_2 = np.random.normal(loc=55, scale=10, size=30)  # 평균 55, 표준편차 10
group_3 = np.random.normal(loc=60, scale=10, size=30)  # 평균 60, 표준편차 10
```

```python
f_stat, p_value = stats.f_oneway(group_1, group_2, group_3)
if p_value < 0.05:
  print(f"p_value: {p_value}, 통계적으로 유의미한 차이")
else:
  print("p_value: {p_value}, 유의미한 차이 X")
```

<aside>
👉

평균이 각각 55와 60이고, 표준편차가 8인 두 개의 그룹(A, B) 데이터를 생성한 후, **두 그룹의 데이터 분포를 Seaborn을 활용하여 KDE(커널 밀도 함수)와 함께 히스토그램으로 시각화**하세요.

이후, **두 그룹 간 평균 차이가 유의미한지 t-검정을 수행하는 코드**를 작성하세요.

</aside>

```python
# 데이터 생성
np.random.seed(42)
group_A = np.random.normal(loc=55, scale=8, size=200)  # 평균 55, 표준편차 8
group_B = np.random.normal(loc=60, scale=8, size=200)  # 평균 60, 표준편차 8
```

```python
import seaborn as sns
plt.figure(figsize=(8, 5))
sns.histplot(group_A, bins=30, alpha=0.5, label="Group_A", kde=True, color='darkorange')
sns.histplot(group_B, bins=30, alpha=0.5, label="Group_B", kde=True, color='purple')
plt.xlabel('Value')
plt.ylabel('Density/Frequency')
plt.show()
```

```python
t_stat, p_value = stats.ttest_ind(group_A, group_B, equal_var=False)
print(f"t-statistic: {t_stat:.4f}, p-value: {p_value:.4f}")
```

이 데이터를 바탕으로 **카이제곱 검정을 수행하여 광고 A와 B의 클릭률 차이가 유의미한지 분석하고, Seaborn의 barplot을 사용하여 클릭률을 비교하는 그래프를 그리는 코드**를 작성하세요.

```python
chi2_stat, p_value, dof, expected = stats.chi2_contingency(observed_data)
print(f"Chi-Square Statistic: {chi2_stat:.4f}. p-value: {p_value:.4f}")
```
