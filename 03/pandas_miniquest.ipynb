````python
import pandas as pd

data = {
    '이름': ['홍길동', '김철수', '박영희', '이순신'],
    '부서': ['영업', '영업', '인사', '인사'],
    '급여': [5000, 5500, 4800, 5100]
}

df = pd.DataFrame(data)
print(df.groupby('부서')['급여'].sum())
````

### **병합 - 미니퀘스트**

1번 퀘스트

<aside>
👉 두 개의 데이터프레임을 `고객ID`를 기준으로 내부 조인(`inner join`)하여 공통된 데이터를 병합하는 코드를 작성하세요.

</aside>

````python
import pandas as pd

df1 = pd.DataFrame({'고객ID': [1, 2, 3], '이름': ['홍길동', '김철수', '이영희']})
df2 = pd.DataFrame({'고객ID': [2, 3, 4], '구매액': [10000, 20000, 30000]})

merged = pd.merge(df1, df2, on='고객ID', how='inner')
print(merged)
````

2번 퀘스트

<aside>
👉 왼쪽 데이터프레임을 기준으로 병합(`left join`)을 수행하고, 구매액이 없는 경우 `NaN`을 유지하는 코드를 작성하세요.

</aside>

````python
import pandas as pd

df1 = pd.DataFrame({'고객ID': [1, 2, 3], '이름': ['홍길동', '김철수', '이영희']})
df2 = pd.DataFrame({'고객ID': [2, 3, 4], '구매액': [15000, 25000, 35000]})
merged = pd.merge(df1, df2, how='left')
print(merged)
````

3번 퀘스트

<aside>
👉 두 개 이상의 열을 기준으로 병합하고, 중복 열의 이름을 구분하도록 접미사(`suffixes`)를 지정하는 코드를 작성하세요.

</aside>

````python
import pandas as pd

df1 = pd.DataFrame({
    '고객ID': [1, 2, 3],
    '도시': ['서울', '부산', '대전'],
    '구매액': [10000, 20000, 30000]
})

df2 = pd.DataFrame({
    '고객ID': [1, 2, 3],
    '도시': ['서울', '부산', '광주'],
    '구매액': [15000, 25000, 35000]
})

merged = pd.merge(df1, df2, on=['고객ID', '도시'], suffixes=('_기존', '_신규'))
print(merged)
````

### **피벗 - 미니퀘스트**

1번 퀘스트

<aside>
👉 주어진 데이터프레임을 `pivot()` 함수를 사용하여 날짜(`날짜`)를 행 인덱스로, 제품(`제품`)을 열로, 판매량(`판매량`)을 값으로 설정하는 코드를 작성하세요.

````python
import pandas as pd

data = {
    '날짜': ['2024-01-01', '2024-01-01', '2024-01-02', '2024-01-02'],
    '제품': ['A', 'B', 'A', 'B'],
    '판매량': [100, 200, 150, 250]
}

df = pd.DataFrame(data)
df_pivot = df.pivot(index='날짜', columns='제품', values='판매량')
print(df_pivot)
````

2번 퀘스트

<aside>
👉 `pivot_table()`을 사용하여 주어진 데이터프레임에서 카테고리(`카테고리`)를 행 인덱스로, 제품(`제품`)을 열로 설정하고, 판매량(`판매량`)의 합계를 출력하는 코드를 작성하세요.

````python
import pandas as pd

data = {
    '카테고리': ['전자', '가전', '전자', '가전'],
    '제품': ['A', 'B', 'A', 'B'],
    '판매량': [100, 200, 150, 250]
}

df = pd.DataFrame(data)
df_pivot = df.pivot_table(index='카테고리', columns='제품', values='판매량', aggfunc='sum')
print(df_pivot)
````

````python
data = pd.read_csv('/content/mymoviedb.csv', engine='python', on_bad_lines='skip')
print(data.head())
print(type(data))
````

````python
# 장르별 평균 인기도
genre_popularity = data.groupby('Genre')[['Popularity']].mean().sort_values(by='Popularity', ascending=False)
print(genre_popularity)
````

````python
# 언어별 영화 개수
language_count = data.groupby('Original_Language').agg(Movie_Count=('Title', 'count')).reset_index().sort_values(by='Movie_Count', ascending=False).reset_index(drop=True)
print(language_count)
````

> 2. 데이터를 DataFrame으로 만들고, 특정 조건을 필터링한 DataFrame을 만드세요.

````python
#평점이 8.0 이상이고 투표 수가 1000개 이상인 영화

data['Vote_Average'] = pd.to_numeric(data['Vote_Average'], errors='coerce')
data['Vote_Count'] = pd.to_numeric(data['Vote_Count'], errors='coerce')
filtered_data = data[(data['Vote_Average'] >= 8.0) & (data['Vote_Count'] >= 1000)]
print(filtered_data)
````
